```python

import pandas as pd

# Read data
sales = pd.read_csv("sales_train_validation.csv")

# Unpivot the last 180 days
sales = sales.melt(
    id_vars=["id", "item_id", "dept_id", "cat_id", "store_id", "state_id"],
    value_vars=sales.columns[-60:],
    var_name="d",
    value_name="units_sold"
)

# Export to CSV
sales.to_csv("sales.csv", index=False)

# 5. Check result
print(sales.shape)
```

```python
calendar = pd.read_csv("calendar.csv")
prices = pd.read_csv("sell_prices.csv")

sales = sales.merge(
    calendar[["d", "date", "wm_yr_wk"]],
    on="d",
    how="left"
)

sales = sales.merge(
    prices,
    on=["item_id", "store_id", "wm_yr_wk"],
    how="left"
)

print(sales.head(10))
```

```python
sales = sales.drop(columns=[
    "date_y", "date",
    "wm_yr_wk_y", "wm_yr_wk",
    "sell_price_y"
])

sales = sales.rename(columns={
    "date_x": "date",
    "wm_yr_wk_x": "wm_yr_wk",
    "sell_price_x": "sell_price"
})
```

```python
print(sales.head(10))
```

```python
sales["revenue"] = sales["units_sold"] * sales["sell_price"]

print(sales.head())
```

```python
sales.to_csv("sales_clean.csv", index=False)
```
